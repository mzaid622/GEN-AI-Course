```python
from langchain_core.tools import tool
from langchain_ollama import ChatOllama
import requests
```

```python
llm = ChatOllama(model="llama3.2:3b")
```

```python
from langchain_community.tools import DuckDuckGoSearchRun

search_tool = DuckDuckGoSearchRun()
```

```python
results = search_tool.invoke("top news in pakistan today")
print(results)
```

```python
from langchain.agents import create_agent
from langchain_classic.agents import AgentExecutor
from langchain_classic import hub
```

```python
# No need to pull a ReAct prompt from the hub anymore.
# create_agent builds a tool-calling agent (via LangGraph) internally,
# so we just give it a plain system_prompt string.
system_prompt = "You are a helpful assistant. Use the search tool when you need up-to-date information."
```

```python
# tools must be a list, and system_prompt is a plain string (not a PromptTemplate)
agent = create_agent(model=llm, tools=[search_tool], system_prompt=system_prompt)
```

```python
# agent_executor = AgentExecutor(
#     agent=agent,
#     tools=[search_tool],
#     verbose=True
# )

# No separate AgentExecutor needed - `agent` is already runnable.
# It's a LangGraph graph, so we invoke it with a "messages" list.
response = agent.invoke(
    {"messages": [{"role": "user", "content": "3 ways to reach goa"}]}
)
print(response)
```

```python
# The final answer is the last message in the returned messages list
response["messages"][-1].content
```
